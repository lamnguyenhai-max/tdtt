```python
print("Ten chu ho:")
name = input()
print("Chi so thang truoc:")
last = int(input())  
print("Chi so thang nay:")
now = int(input()) 

n = now - last
sum = 0

if n > 400:
    sum += (n - 400) * 3460
    n = 400
if n > 300:
    sum += (n - 300) * 3350
    n = 300
if n > 200:
    sum += (n - 200) * 2998
    n = 200
if n > 100:
    sum += (n - 100) * 2380
    n = 100
if n > 50:
    sum += (n - 50) * 2050
    n = 50
if n > 0:
    sum += n * 1984
final_sum = round(sum * 1.08)

print(f"Ho va ten: {name}")
print(f"Tien phai tra la: {final_sum}")
```
