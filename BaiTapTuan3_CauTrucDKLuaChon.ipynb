W3A16:

```python
def check(x) :
    i = int(x)
    if(x == i) : c = f = i
    elif(x > 0) : c, f = i + 1, i
    else : c, f = i, i - 1
    
    if(x - f >= 0.5) : n = c
    else : n = f
    
    return f"{c} {f} {n}"

x = float(input())
print(check(x))
```

W3A17:

```python
def check(a, b, c, d) :
    if(a != 0 and b % a == 0) :
        q = b // a
        if(b * q == c and c * q == d) : return "YES"
    if(a == b == c == d == 0) : return "YES"
    return "NO"

a, b, c, d = map(int, input().split())
print(check(a, b, c, d))
```
