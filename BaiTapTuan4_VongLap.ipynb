```python
import math

a, b = map(int, input().split())

res = 0

for num in range(a, b + 1):
    if num >= 2:
        c = True
        for i in range(2, int(math.sqrt(num)) + 1):
            if num % i == 0:
                c = False
                break
        if c:
            res += num

print(res)
```

W4A8:

```python
n = int(input())

steps = 0
res = n

while True:
    temp = res
    rev = 0
    while temp > 0:
        rev = rev * 10 + temp % 10
        temp //= 10
    if res == rev:
        break
    res = res + rev
    steps += 1

print(steps, res)
```

W4A9:

```python
n = int(input())

res = []
k = 1

while k * k <= n:
    sq = k * k
    temp = sq
    seen = [False] * 10
    ok = True

    while temp > 0:
        d = temp % 10
        if seen[d]:
            ok = False
            break
        seen[d] = True
        temp //= 10

    if ok:
        res.append(str(sq))
    k += 1

if len(res) == 0:
    print("(no number)")
else:
    print(" ".join(res))
```

W4A13:

```python
import math

a = int(input())
b = int(input())

sa = 1 if a > 1 else 0
for i in range(2, int(math.sqrt(a)) + 1):
    if a % i == 0:
        sa += i
        j = a // i
        if j != i:
            sa += j

sb = 1 if b > 1 else 0
for i in range(2, int(math.sqrt(b)) + 1):
    if b % i == 0:
        sb += i
        j = b // i
        if j != i:
            sb += j

print("true" if sa == b and sb == a else "false")
```
